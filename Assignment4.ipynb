## Q1

```python
import pandas as pd

fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.", "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.", "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.", "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.", "keywords": "pay payment upi fee", "category": "billing"}
]
roll_number = 1024170260
last_digit_1 = 6
last_digit_2 = 0
categories = ["billing", "account", "general"]
category_1 = categories[last_digit_1 % 3]
category_2 = categories[last_digit_2 % 3]
personal_entry_1 = {
    "question": "how can i pay my bill",
    "answer": "You can pay your bill using UPI, card, or net banking.",
    "keywords": "bill payment pay upi card",
    "category": category_1
}
personal_entry_2 = {
    "question": "how can i check my billing details",
    "answer": "You can check your billing details from the billing section.",
    "keywords": "billing details bill payment amount",
    "category": category_2
}
all_entries = fixed_entries + [personal_entry_1, personal_entry_2]
df = pd.DataFrame(all_entries)
print(df)
```

## Q2

```python
def score_query(query, df):
    query_words = query.lower().split()
    results = []
    for i in range(len(df)):
        keywords = df.iloc[i]['keywords'].lower().split()
        score = 0
        for word in query_words:
            if word in keywords:
                score = score + 1
        if score > 0:
            confidence = score / len(query_words)
            results.append([i, df.iloc[i]['question'], score, confidence])
    result_df = pd.DataFrame(results, columns=['Index', 'Question', 'Score', 'Confidence'])
    result_df = result_df.sort_values(by='Confidence', ascending=False)
    return result_df
```

```python
query = "pay fee"
result = score_query(query, df)
print(result)
```

## Q6

```python
def score_query_with_ties(query, df):
    query_words = query.lower().split()
    results = []

    for i in range(len(df)):
        keywords = df.iloc[i]['keywords'].lower().split()
        score = 0

        for word in query_words:
            if word in keywords:
                score = score + 1

        if score > 0:
            confidence = score / len(query_words)
            results.append([i, df.iloc[i]['question'], score, confidence])

    result_df = pd.DataFrame(results, columns=['Index', 'Question', 'Score', 'Confidence'])

    if len(result_df) == 0:
        print("No matching FAQ found.")
        return result_df

    highest_score = result_df['Score'].max()

    best_matches = result_df[result_df['Score'] == highest_score]

    print("All best matching entries:")
    print(best_matches)

    return best_matches
```

```python
query = "fee"
result = score_query_with_ties(query, df)
print(result)
```
