## Q1

```python
#(i)
roll_no="1024170260"
L=[int(digit)*10 for digit in roll_no]
print("L: ",L)
```

```python
#(ii)
# append() adds the element at the end
L.append(90)
print("Using append: ",L)
# insert() adds the element at a specified index
L.insert(3,55)
print("Using insert: ",L)
```

```python
#(iii)
L.remove(10)
print("L after removing 10: ",L)
L.pop(2)
print("L after removing element at index 2: ",L)
```

```python
#(iv)
L.sort()
print("Ascending order: ",L)
L.sort(reverse=True)
print("Descending order: ",L)
```

```python
#(v)
print("First three elements: ",L[:3])
print("Last three elements: ",L[-3:])
```

```python
#(vi)
average_L=sum(L)/len(L)
print("Average: ",average_L)
new_list=[x for x in L if x>average_L]
print("Elements greater than average: ",new_list)
```

## Q2

```python
scores=tuple(L[:8])
print("Original scores tuple: ",scores)
#(i)
highest_score=max(scores)
highest_idx=scores.index(highest_score)
lowest_score=min(scores)
lowest_count=scores.count(lowest_score)
print("Highest Score: ",highest_score," index: ",highest_idx)
print("Lowest Score: ",lowest_score," count: ",lowest_count)
```

```python
#(ii)
reversed_list=list(scores)[::-1]
print(reversed_list)
# Tuples can't be reversed in-place because they are immutable (unchangeable after creation)
```

```python
#(iii)
user_score=int(input("Enter a number to find: "))
if user_score in scores:
    print(user_score," has first presence at index: ",scores.index(user_score))
else:
    print(user_score, " is not present at any index")
```

```python
#(iv)
try:
    scores[0]=100
except TypeError:
    print("Tuples are immutable (read-only), unlike mutable lists which allow direct item reassignment.")
```

```python
#(v)
first,second,*remaining=scores
print("First score: ",first)
print("Second score: ",second)
print("Remaining scores: ",remaining)
```

## Q3

```python
#(i)
import random
random.seed(1024170260)
L=[random.randint(100,900) for i in range(100)]
L
```

```python
#(ii)
cnt=0
for val in L:
    if val%2==1:
        cnt+=1
print("No. of odd numbers: ",cnt)
```

```python
#(iii)
cnt=0
for val in L:
    if val%2==0:
        cnt+=1
print("No. of even numbers: ",cnt)
```

```python
#(iv)
def checkPrime(n):
    if n<2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True
primes=[num for num in L if checkPrime(num)]
print("Count of primes: ",len(primes))
print("Primes: ",primes)
```

```python
#(v)
freq={}
for num in L:
    if num in freq:
        freq[num]+=1
    else:
        freq[num]=1
most_freq=0
max_count=0
for key in freq:
    if freq[key]>max_count:
        max_count=freq[key]
        most_freq=key
print("Most frequent element: ",most_freq)
print("Count of most frequent element: ",max_count)
```

## Q4

```python
roll_no="1024170260"
digits=[int(d) for d in roll_no[:8]]
A={d*7 for d in digits}
B={d*9 for d in digits}
print("Set A: ",A)
print("Set B: ",B)
```

```python
#(i)
union_set=A.union(B)
print("Union of A and B: ",union_set)
```

```python
#(ii)
intersection_set=A.intersection(B)
print("Intersection of A and B: ", intersection_set)
```

```python
#(iii)
A_diff_B=A.difference(B)
B_diff_A=B.difference(A)
print("A-B: ",A_diff_B)
print("B-A: ",B_diff_A)
# difference() return elements in one set that are not in the other, whereas symmetric_difference() returns all elements that are not shared in both sets combined.
```

```python
#(iv)
symm_diff=A.symmetric_difference(B)
print("Symmetric difference of A and B: ", symm_diff)
```

```python
#(v)
A_subset_B=A.issubset(B)
B_superset_A=B.issuperset(A)
print("Is A subset of B? ",A_subset_B)
print("Is B superset of A? ",B_superset_A)
```

```python
#(vi)
x=int(input("Enter a value to remove from A"))
A.discard(x)
print("Set A after discard: ",A)
#discard() is safer than remove() because if the value doesn't exist in the set, discard() does nothing quietly, whereas remove() crashes the program with a KeyError.
```
