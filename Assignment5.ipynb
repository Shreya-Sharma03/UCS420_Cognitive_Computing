## Q.2 Questions on Basic NumPy Array: 

## a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5]) 

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("Reversed Array:", reversed_arr)
```

## b) Find the most frequent value and their indice(s) in the following arrays: 
## i. x = np.array([1,2,3,4,5,1,2,1,1,1]) 

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = vals_x[np.argmax(counts_x)]
indices_x = np.where(x == most_freq_x)[0]
print(f"Most frequent in x: {most_freq_x}, Indices: {indices_x}")
```

## ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ]) 

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals_y, counts_y = np.unique(y, return_counts=True)
most_freq_y = vals_y[np.argmax(counts_y)]
indices_y = np.where(y == most_freq_y)[0]
print("Most frequent in y:", most_freq_y, ", Indices:", indices_y)
```

## Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute? 

```python
import numpy as np
shreya_sharma = np.linspace(10, 100, 25)
print("Dimensions (ndim):", shreya_sharma.ndim)
print("Shape:", shreya_sharma.shape)
print("Total elements (size):", shreya_sharma.size)
print("Data type (dtype):", shreya_sharma.dtype)
print("Bytes consumed (nbytes):", shreya_sharma.nbytes)
transposed_reshape = shreya_sharma.reshape(25, 1) 
print(shreya_sharma.reshape(1, 25))
print("\nCan we use .T attribute directly on 1-D array?", shreya_sharma.T)
```

## Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>. 

```python
ucs420_shreya_sharma = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])
print(np.mean(ucs420_shreya_sharma))
print(np.median(ucs420_shreya_sharma))
print(np.max(ucs420_shreya_sharma))
print(np.min(ucs420_shreya_sharma))
print(np.unique(ucs420_shreya_sharma))
reshaped_ucs420_shreya_sharma = ucs420_shreya_sharma.reshape(4, 3)
print(reshaped_ucs420_shreya_sharma)
resized_ucs420_shreya_sharma = np.resize(ucs420_shreya_sharma, (2, 3))
print(resized_ucs420_shreya_sharma)
```
