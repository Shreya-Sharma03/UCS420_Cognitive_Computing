## Q6

```python
data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data)
employees.to_csv('employees.csv', index=False)
print(employees)
```

```python
# (a)
print(employees.shape)
```

```python
# (b)
print(employees.info())
```

```python
# (c)
print(employees.describe())
```

```python
# (d)
print(employees.head(5))
print(employees.tail(3))
```

```python
# (e-i)
print(employees['Salary'].mean())
```

```python
# (e-ii)
print(employees['Bonus'].sum())
```

```python
# (e-iii)
print(employees['Age'].min())
```

```python
# (e-iv)
print(employees['Rating'].max())
```

```python
# (f)
employees = employees.sort_values(by='Salary', ascending=False)
print(employees)
```

```python
# (g)
employees['Performance'] = employees['Rating'].apply(
    lambda x: 'Excellent' if x >= 4.5 else ('Good' if x >= 4.0 else 'Average')
)
print(employees)
```

```python
# (h)
print(employees.isnull().sum())
```

```python
# (i)
employees = employees.rename(columns={'Employee_ID': 'ID'})
print(employees)
```

```python
# j) i.
print(employees[employees['Years_of_Experience'] > 5])
```

```python
# j) ii.
print(employees[employees['Department'] == 'IT'])
```

```python
# (k)
employees['Tax'] = employees['Salary'] * 0.10
print(employees)
```

```python
# (l)
employees.to_csv('modified_employees.csv', index=False)
print("Modified DataFrame saved successfully.")
```
